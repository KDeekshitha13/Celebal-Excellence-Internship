```python
import random
from faker import Faker
from datetime import datetime, timedelta
import pandas as pd

fake = Faker()

random.seed(42)
Faker.seed(42)
```

```python
NUM_CUSTOMERS = 500
NUM_PRODUCTS = 500
NUM_ORDERS = 500
NUM_ORDER_ITEMS = 1000
```

```python
customer_types = ["REGULAR", "PREMIUM", "VIP"]

customers = []

for i in range(1, NUM_CUSTOMERS + 1):

    customers.append({
        "customer_id": i,
        "customer_name": fake.name(),
        "email": fake.email(),
        "registration_date": fake.date_between(
            start_date="-3y",
            end_date="today"
        ),
        "customer_type": random.choice(customer_types)
    })
```

```python
num_invalid = int(NUM_CUSTOMERS * 0.02)
indices = random.sample(range(NUM_CUSTOMERS), num_invalid)
for i in indices:
    customers[i]["email"] = customers[i]["email"].replace("@", "")
```

```python
customers_df = pd.DataFrame(customers)
customers_df.head()
```

```python
categories = {

    "Electronics": [
        "Laptop",
        "Phone",
        "Camera",
        "Headphones",
        "Monitor"
    ],

    "Clothing": [
        "Shirt",
        "Jeans",
        "Jacket",
        "Dress",
        "T-Shirt"
    ],

    "Books": [
        "Novel",
        "Biography",
        "Dictionary",
        "Magazine",
        "Comics"
    ],

    "Home": [
        "Chair",
        "Table",
        "Lamp",
        "Fan",
        "Sofa"
    ]
}
```

```python
products = []
for i in range(1, NUM_PRODUCTS + 1):
    category = random.choice(list(categories.keys()))
    product = random.choice(categories[category])
    products.append({
        "product_id": i,
        "product_name": product,
        "category": category,
        "subcategory": product,
        "cost_price": round(
            random.uniform(100, 5000), 2
        )
    })
```

```python
num_dirty = int(NUM_PRODUCTS * 0.03)

indices = random.sample(range(NUM_PRODUCTS), num_dirty)

for i in indices:

    choice = random.randint(1,3)

    if choice == 1:
        products[i]["product_name"] = "  " + products[i]["product_name"] + "  "

    elif choice == 2:
        products[i]["product_name"] = products[i]["product_name"].upper()

    else:
        products[i]["product_name"] = products[i]["product_name"].lower()
```

```python
products_df = pd.DataFrame(products)

products_df.head()
```

```python
statuses = [
    "PLACED",
    "SHIPPED",
    "DELIVERED",
    "CANCELLED",
    "RETURNED"
]

regions = [
    "NORTH",
    "SOUTH",
    "EAST",
    "WEST"
]
```

```python
orders = []

for i in range(1, NUM_ORDERS + 1):

    dt = fake.date_time_between(
        start_date="-2y",
        end_date="now"
    )

    orders.append({

        "order_id": i,

        "customer_id": random.randint(1, NUM_CUSTOMERS),

        "order_date": dt.strftime("%Y-%m-%d %H:%M:%S"),

        "status": random.choice(statuses),

        "region_code": random.choice(regions)

    })
```

```python
num_null = int(NUM_ORDERS * 0.05)

indices = random.sample(range(NUM_ORDERS), num_null)

for i in indices:

    orders[i]["customer_id"] = None
```

```python
num_wrong = int(NUM_ORDERS * 0.03)

indices = random.sample(range(NUM_ORDERS), num_wrong)

for i in indices:

    dt = datetime.strptime(
        orders[i]["order_date"],
        "%Y-%m-%d %H:%M:%S"
    )

    orders[i]["order_date"] = dt.strftime("%d-%m-%Y")
```

```python
orders_df = pd.DataFrame(orders)
```

```python
order_items = []

for i in range(1, NUM_ORDER_ITEMS + 1):

    order_items.append({

        "item_id": i,

        "order_id": random.randint(1, NUM_ORDERS),

        "product_id": random.randint(1, NUM_PRODUCTS),

        "quantity": random.randint(1,5),

        "unit_price": round(
            random.uniform(100,5000),2
        ),

        "discount_percent": random.randint(0,50)

    })
```

```python
num_negative = int(NUM_ORDER_ITEMS * 0.03)

indices = random.sample(range(NUM_ORDER_ITEMS), num_negative)

for i in indices:

    order_items[i]["quantity"] = -random.randint(1,3)
```

```python
num_invalid = int(NUM_ORDER_ITEMS * 0.02)

indices = random.sample(range(NUM_ORDER_ITEMS), num_invalid)

for i in indices:

    order_items[i]["order_id"] = 999999
```

```python
order_items_df = pd.DataFrame(order_items)
```

```python
customers_df.to_csv("customers.csv", index=False)

products_df.to_csv("products.csv", index=False)

orders_df.to_csv("orders.csv", index=False)

order_items_df.to_csv("order_items.csv", index=False)
```

```python
customers = pd.read_csv("customers.csv")

products = pd.read_csv("products.csv")

orders = pd.read_csv("orders.csv")

order_items = pd.read_csv("order_items.csv")
```

```python
print(customers.isnull().sum())

print(products.isnull().sum())

print(orders.isnull().sum())

print(order_items.isnull().sum())
```

```python
customers.drop_duplicates(inplace=True)

products.drop_duplicates(inplace=True)

orders.drop_duplicates(inplace=True)

order_items.drop_duplicates(inplace=True)
```

```python
customers["customer_name"] = customers["customer_name"].fillna("Unknown")
```

```python
pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
```

```python
customers["valid_email"] = customers["email"].str.match(pattern, na=False)
```

```python
invalid_emails = customers[customers["valid_email"] == False]

print(invalid_emails)
```

```python
products["product_name"] = products["product_name"].str.strip()
```

```python
products["product_name"] = products["product_name"].str.title()
```

```python
products.drop_duplicates(subset="product_id", inplace=True)
```

```python
orders["customer_id"] = orders["customer_id"].fillna(0)
```

```python
orders["order_date"] = pd.to_datetime(

    orders["order_date"],

    errors="coerce",

    dayfirst=True

)
```

```python
orders["order_date"] = orders["order_date"].dt.strftime("%Y-%m-%d %H:%M:%S")
```

```python
orders = orders.dropna(subset=["order_date"])
```

```python
order_items = order_items[order_items["quantity"] > 0]
```

```python
order_items = order_items[
    order_items["discount_percent"] <= 100
]
```

```python
invalid_orders = order_items[
    ~order_items["order_id"].isin(
        orders["order_id"]
    )
]
```

```python
print(invalid_orders)
```

```python
order_items = order_items[
    order_items["order_id"].isin(
        orders["order_id"]
    )
]
```

```python
invalid_products = order_items[
    ~order_items["product_id"].isin(
        products["product_id"]
    )
]

print(invalid_products)
```

```python
invalid_products = order_items[
    ~order_items["product_id"].isin(
        products["product_id"]
    )
]

print(invalid_products)
```

```python
customers.to_csv(
    "customers_clean.csv",
    index=False
)

products.to_csv(
    "products_clean.csv",
    index=False
)

orders.to_csv(
    "orders_clean.csv",
    index=False
)

order_items.to_csv(
    "order_items_clean.csv",
    index=False
)
```
