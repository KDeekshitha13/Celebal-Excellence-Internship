```python
orders_df = spark.table("orders")
order_items_df = spark.table("order_items")
```

```python
def test_invalid_order_id():
    invalid = order_items_df.join(
        orders_df,
        order_items_df.order_id == orders_df.order_id,
        "left_anti"
    )

    count = invalid.count()

    if count == 0:
        print("✅ PASS: All order_ids are valid.")
    else:
        print(f"❌ FAIL: {count} invalid order_id(s) found.")
        invalid.show()
```

```python
test_invalid_order_id()
```

```python
def test_discount_percent():
    invalid = order_items_df.filter("discount_percent > 100")

    count = invalid.count()

    if count == 0:
        print("✅ PASS: No discount greater than 100%.")
    else:
        print(f"❌ FAIL: {count} invalid discount(s).")
        invalid.show()
```

```python
test_discount_percent()
```

```python
def test_zero_quantity():
    invalid = order_items_df.filter("quantity = 0")

    count = invalid.count()

    if count == 0:
        print("✅ PASS: No zero quantities.")
    else:
        print(f"❌ FAIL: {count} rows with quantity = 0.")
        invalid.show()
```

```python
test_zero_quantity()
```

```python
from pyspark.sql.functions import current_date, to_date

def test_future_orders():
    invalid = orders_df.filter(
        to_date("order_date") > current_date()
    )

    count = invalid.count()

    if count == 0:
        print("✅ PASS: No future order dates.")
    else:
        print(f"❌ FAIL: {count} future order(s) found.")
        invalid.show()
```

```python
test_future_orders()
```

```python
print("Running Edge Case Tests...\n")

test_invalid_order_id()
test_discount_percent()
test_zero_quantity()
test_future_orders()

print("\nAll tests completed.")
```
