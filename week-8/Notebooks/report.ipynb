```python
report_type = input("Enter report type (daily/weekly/monthly): ").strip().lower()

start_date = input("Enter start date (YYYY-MM-DD): ")
end_date = input("Enter end date (YYYY-MM-DD): ")
```

```python
orders = spark.sql(f"""
SELECT COUNT(*) AS total_orders
FROM orders
WHERE DATE(order_date)
BETWEEN '{start_date}' AND '{end_date}'
""")

orders.show()
```

```python
revenue = spark.sql(f"""
SELECT
ROUND(SUM(quantity*unit_price*(1-discount_percent/100)),2) AS total_revenue
FROM order_items oi
JOIN orders o
ON oi.order_id=o.order_id
WHERE DATE(order_date)
BETWEEN '{start_date}' AND '{end_date}'
""")

revenue.show()
```

```python
customers = spark.sql(f"""
SELECT COUNT(DISTINCT customer_id) AS unique_customers
FROM orders
WHERE DATE(order_date)
BETWEEN '{start_date}' AND '{end_date}'
""")

customers.show()
```

```python
top_products = spark.sql(f"""
SELECT
p.product_name,
SUM(quantity) AS total_quantity
FROM order_items oi
JOIN products p
ON oi.product_id=p.product_id
JOIN orders o
ON oi.order_id=o.order_id
WHERE DATE(order_date)
BETWEEN '{start_date}' AND '{end_date}'
GROUP BY p.product_name
ORDER BY total_quantity DESC
LIMIT 3
""")

top_products.show()
```

```python
previous = spark.sql(f"""
SELECT
ROUND(SUM(quantity*unit_price*(1-discount_percent/100)),2) AS previous_revenue
FROM order_items oi
JOIN orders o
ON oi.order_id=o.order_id
WHERE DATE(order_date)
<
'{start_date}'
""")

previous.show()
```

```python
print("="*40)
print("ORDER SUMMARY REPORT")
print("="*40)

orders.show()

revenue.show()

customers.show()

print("Top Products")
top_products.show()

print("Previous Revenue")
previous.show()
```
